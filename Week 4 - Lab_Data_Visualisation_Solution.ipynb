## CASE DESCRIPTION:

This case study investigates the factors that contribute to student performance in final exams in HCT.  The data for this study was collected randomly from 30 students’ performance in their exams spread around various departments and campuses. <br><br>
This case study was conducted to get some insights into the factors that can affect student success in final exams visually through exploratory analysis. By understanding these factors, educators can develop strategies to help students improve their performance.

```python
import pandas as pd
df = pd.read_csv('Data/MarksHCT.csv')
df
```

```python
df.info()
```

```python
df['Final'] = df['Final'].astype(float)
df
```

```python
# Fill NaN in "Quiz" with the overall average of the Quiz column
df["Quiz"].fillna(df["Quiz"].mean(), inplace=True)

# or fill with the mean of that particular student mark. 
#df["Quiz"].fillna(df[["Quiz", "Practical", "Project", "Final"]].mean(axis=1),inplace=True)

df
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Suppose your dataframe is called df
# Columns: ID, Division, Quiz, Practical, Project, Final

# Melt the dataframe so we can plot multiple exam types easily
df_melted = df.melt(id_vars=["ID", "Division"], 
                    value_vars=["Quiz", "Project", "Final"], 
                    var_name="Exam", value_name="Marks")

# Draw boxplot
plt.figure(figsize=(10,6))
sns.boxplot(x="Division", y="Marks", hue="Exam", data=df_melted, palette="Set2")

plt.title("Division-wise Performance in Quiz, Project, and Final")
plt.xlabel("Division")
plt.ylabel("Marks")
plt.legend(title="Exam")

# --- Save the figure as an image you can download ---
import os
from IPython.display import FileLink

# Make sure layout isn't clipped
plt.tight_layout()

# Choose a filename (change if you like)
out_png = "division_performance.png"
out_pdf = "division_performance.pdf"   # vector (great for print)

# Save (PNG raster + PDF vector)
plt.savefig(out_png, dpi=300, bbox_inches="tight")
plt.savefig(out_pdf, bbox_inches="tight")   # vector, no dpi needed

# Optional: transparent background
# plt.savefig(out_png, dpi=300, bbox_inches="tight", transparent=True)

# Show clickable download links in the notebook
display(FileLink(out_png))
display(FileLink(out_pdf))

# If you still want to display the plot in the cell:
plt.show()


plt.show()
```

## Create a heat map for number of students gender wise for each zone

```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt

# load the data into a pandas dataframe
df = pd.read_csv('Data/Students.csv')

# create a pivot table with Zone as rows, Gender as columns, and student count as values
pivot_table = pd.pivot_table(df, index='Zone_EN', values='Student_Count', aggfunc='sum')

# create a heatmap
sns.heatmap(pivot_table, cmap='Reds')

# add labels and title
plt.ylabel('Zone')
plt.title('Total Number of Students by Zone')

# display the plot
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# load the data into a pandas dataframe
df = pd.read_csv('Data/Students.csv')

# create a pivot table with Zone as rows and total student count as values
pivot_table = pd.pivot_table(df, index='Zone_EN', values='Student_Count', aggfunc='sum')

# sort zones from largest to smallest
pivot_table = pivot_table.sort_values(by='Student_Count', ascending=False)

# create a heatmap
sns.heatmap(pivot_table, cmap='Blues')

# add labels and title
plt.ylabel('Zone')
plt.title('Total Number of Students by Zone')

# display the plot
plt.show()
```

## Create a heat map for number of students by dividing into three segments

```python
# libraries
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# create data
df = pd.read_csv('Data/Students.csv')

# make the values in dataset discrete
# the values will be cut into 3 discrete values: 0,1,2
df_q = pd.DataFrame()
for col in df:
    df_q['Student_Count']= pd.to_numeric(pd.qcut(df['Student_Count'], 3, labels=list(range(3))))

# plot it
sns.heatmap(df_q)
plt.show()
```

## Create a heat map for Number of Students by Zone and Gender

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the CSV file
df = pd.read_csv('Data/Students.csv')

# Creating a pivot table to summarize data by Zone and Gender
heatmap_data = df.pivot_table(
    index='Zone_EN',
    columns='Gender_EN',
    values='Student_Count',
    aggfunc='sum'
)

# Plotting the heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(heatmap_data, annot=True, fmt=".0f", cmap="coolwarm", cbar_kws={'label': 'Number of Students'})
plt.title('Number of Students by Zone and Gender')
plt.ylabel('Zone')
plt.xlabel('Gender')
plt.show()

```
